### Forward Filling (`ffill`) and Backward Filling (`bfill`) ---> Time Series Data

#### **Forward Filling (`ffill`) → Fill missing values using the previous available value.**

#### **Backward Filling (`bfill`) → Fill missing values using the next available value.**

### Example

**Original:**

`10 → NaN → NaN → 20`

**Forward Fill:**

`10 → 10 → 10 → 20`

**Backward Fill:**

`10 → 20 → 20 → 20`

### Pandas Syntax

```python
# Forward Fill
df['column'].ffill()

# Backward Fill
df['column'].bfill()
